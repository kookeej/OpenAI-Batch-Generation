```python
import json
from pathlib import Path
```

```python
# Set data path
data_path = Path('your/data/path')

data = []
with open(data_path / 'your_data_name.jsonl', 'r') as f:
    for line in f:
        data.append(json.loads(line))

print("# data:", len(data))
```

```python
template = """YOUR TEMPLATE
"""
```

```python
# input file format
{
    "custom_id": "unique_id",
    "method": "POST",
    "url": "/v1/chat/completions",
    "body": {"model": "gpt-4o-mini-2024-07-18", "max_tokens": 300, "temperature": 0.0,  "messages": [{"role": "user", "content": "input_prompt"}]}
}
```

```python
for i in range(len(data)):
    input_prompt = template.format(
        input_args=""
    )
    
    with open(f'../batch_input/batch_input_your_data_name.jsonl', 'a') as f:
        json.dump({
            "custom_id": data[i]['id'],
            "method": "POST",
            "url": "/v1/chat/completions",
            "body": {
                "model": 'gpt-4o-mini',
                "temperature": 0.0,
                "messages": [{"role": "user", "content": input_prompt}]
            }
        }, f)
        f.write('\n')
```
